```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
# Load merged AQI dataset
df = pd.read_csv("../data/processed/merged_aqi.csv")

# Display first 5 rows
df.head()
```

```python
print("=" * 60)
print("Dataset Shape")
print("=" * 60)

print(df.shape)

print("\n")

print("=" * 60)
print("Dataset Information")
print("=" * 60)

df.info()
```

```python
print("=" * 60)
print("Missing Values")
print("=" * 60)

missing = df.isnull().sum()

print(missing)

print("\nPercentage Missing")

print((missing / len(df) * 100).round(2))
```

```python
duplicates = df.duplicated().sum()

print(f"Duplicate Rows: {duplicates}")

df.drop_duplicates(inplace=True)

print("Dataset Shape After Removing Duplicates:")
print(df.shape)
```

```python
df["Date"] = pd.to_datetime(df["Date"])

print(df["Date"].dtype)
```

```python
pollutants = [
    "PM2.5",
    "PM10",
    "Ozone",
    "SO2",
    "NO2"
]

for col in pollutants:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print(df[pollutants].dtypes)
```

```python
print(df[pollutants].isnull().sum())
```

```python
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day
df["DayOfWeek"] = df["Date"].dt.day_name()

df.head()
```

```python
df.describe(include="all")
```

```python
output_path = "../data/processed/cleaned_aqi.csv"

df.to_csv(output_path, index=False)

print(f"Cleaned dataset saved to:\n{output_path}")
```
